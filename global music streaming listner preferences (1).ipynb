```python
#Segment with highest Premium-user percentage
 

result = (
    df.assign(Premium=df['Subscription'] == 'Premium')
      .groupby('Engagement_Level')['Premium']
      .mean()
      .mul(100)
      .sort_values(ascending=False)
)

print(result)
```

```python
#Country-wise genre popularity
result=pd.crosstab(df['Country'],df['Favorite_Genre'])
print(result)
```

```python
#country-wise average listening hours and spending using agg()
result=df.groupby('Country').agg(avg_listening_hours=('Listening_Hours_Per_Week','mean'),average_spending=('Monthly_Spend_USD','mean'))
print(result)       
```

```python
#Subscription-wise average listening time and songs played
result=df.groupby('Subscription').agg(avg_listening_hours=('Listening_Hours_Per_Week','mean'),average_songs_played=('Songs_Played_Per_Day','mean'))
print(result)
```

```python
#Genre-wise total listeners and average listening time
result=df.groupby('Favorite_Genre').agg(total_listener=('Listener_ID','count'),average_listening_time=( 'Listening_Hours_Per_Week','mean'))
print(result)
```

```python
#Device-wise average listening time and spending
result=df.groupby('Preferred_Device').agg(average_listening_time=('Listening_Hours_Per_Week','mean'),average_spendig_time=( 'Monthly_Spend_USD','mean'))
print(result)
```
